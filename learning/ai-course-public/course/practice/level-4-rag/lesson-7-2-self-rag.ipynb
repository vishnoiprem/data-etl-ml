## Concept (5 min)\n\nSelf-RAG is the pattern where the **model critiques its own retrieval** and decides what to do next. The flow: (1) Retrieve docs. (2) Model says: are these relevant? (yes/no/partial). (3) If no, rewrite the query and retrieve again. (4) If partial, retrieve more. (5) If yes, generate the answer. (6) Model critiques the answer: is it grounded in the docs? (yes/no). (7) If no, regenerate. **CRAG** (Corrective RAG) is a simpler version: always evaluate retrieval, replace with web search if irrelevant, refine if partially relevant. **Adaptive retrieval**: only retrieve when the query needs it (e.g., 'hello' doesn't need retrieval). **Architect trade-off**: Self-RAG adds 2-3 extra LLM calls per query (3-5x cost). Worth it for high-stakes apps (medical, legal, financial).\n
