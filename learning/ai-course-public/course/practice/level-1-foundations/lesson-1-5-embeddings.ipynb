## Concept (5 min)\n\nAn **embedding** is a list of 384-3072 floats that represents the meaning of a piece of text. Two pieces of text with similar meaning have vectors that point in similar directions. Cosine similarity measures how parallel two vectors are; values close to 1.0 mean 'semantically equivalent'. Embeddings are the foundation of search, RAG, clustering, recommendations, and classification. The common models are OpenAI's `text-embedding-3-small` (1536 dims, $0.02/1M tokens) and `text-embedding-3-large` (3072 dims, $0.13/1M tokens). Open-source alternatives are `BGE-small` (384 dims, free, local) and `Cohere embed-v3`. The dimension matters: 384 dims is fast but less accurate; 3072 dims is accurate but expensive to store and slow to search. For most apps, 1536 dims (OpenAI small) is the sweet spot. Visualization uses PCA or t-SNE to project the high-dimensional vectors into 2D for inspection.\n
