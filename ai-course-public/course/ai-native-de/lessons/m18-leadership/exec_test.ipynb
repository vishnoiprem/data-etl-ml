## Plain English (the 12-year-old version)

A **STAR story** is a 4-part answer to a behavioral interview question: **Situation** (where you were), **Task** (what you owned), **Action** (what you actually did), **Result** (the number). A **STAR library** is 8-12 of those stories, organized so you can find the right one in 30 seconds and rehearse it three different ways.

The four leadership principles (used by META, Google, and most big tech) are:

1. **Vision** -- set direction, write the roadmap
2. **Execution** -- ship on time, run incidents
3. **Collaboration** -- cross-team alignment
4. **Technical depth** -- system design, debugging

Plus a 5th: **Mentorship** -- grow others.

The **3-2-1 rehearsal framework** says every story needs three layers of compression: a 3-minute long-form talk, a 2-minute deep-dive on the most interesting decision, and a 1-minute Q&A version. The 5 story banks (RAG, cost, security, scale, hiring) are the categories of work most staff candidates should be able to anchor at least one story in.

The lesson's point: **strong candidates maintain the library for years; weak ones try to improvise.**

## The 5 things to know

1. **STAR = Situation, Task, Action, Result** -- the canonical behavioral answer format. Action is first-person; Result is a number.
2. **The 4 (+1) leadership principles** -- Vision, Execution, Collaboration, Tech Depth, plus Mentorship. Most big tech loops score on these.
3. **The 5 story banks** -- RAG, cost, security, scale, hiring. Have at least one story per bank before your loop.
4. **The 3-2-1 framework** -- 3-min talk, 2-min deep dive, 1-min Q&A. Every story gets all three.
5. **The library is the artifact** -- writing stories down (with a heuristic scorer) is the difference between a 5/10 and an 8/10 loop.

````python
# === Section 1: Library overview ===
library = _L18._build_sample_library()
print(f'Library has {len(library)} stories\n')
print('Bank distribution:')
from collections import Counter
banks = Counter(s.bank or '(no bank)' for s in library.stories)
for bank, n in sorted(banks.items()):
    print(f'  {bank:<10}  {n} stories')
print('\nStory titles:')
for s in library.stories:
    print(f'  - {s.title}  [{s.bank or "no bank"}]')
````

````python
# === Section 2: Principle coverage per story ===
print(f'{"story":<48}  {"principles covered":<60}')
print('-' * 110)
for s in library.stories:
    ps = [p.value for p in _L18.principles_covered(s)]
    print(f'{s.title[:48]:<48}  {", ".join(ps):<60}')
````

````python
# === Section 3: Heuristic story scoring ===
print(f'{"title":<48}  {"score":>5}  {"quant":>5} {"act":>4} {"imp":>4} {"dur":>4} {"role":>5}')
print('-' * 90)
for s in library.stories:
    sc = _L18.score_story(s)
    flags = (
        f"{'Y' if sc.result_quantified else 'N':>5} "
        f"{'Y' if sc.action_clear else 'N':>4} "
        f"{'Y' if sc.impact_real else 'N':>4} "
        f"{'Y' if sc.duration_ok else 'N':>4} "
        f"{'Y' if sc.role_aligned else 'N':>5}"
    )
    print(f'{s.title[:48]:<48}  {sc.total:>5}  {flags}')
````

````python
# === Section 4: match_stories for hard questions ===
sample_qs = [
    _L18.MOCK_QUESTIONS[6],   # migration under deadline
    _L18.MOCK_QUESTIONS[8],   # deliver bad news
    _L18.MOCK_QUESTIONS[18],  # 100M-doc retrieval design
    _L18.MOCK_QUESTIONS[25],  # person you grew
    _L18.MOCK_QUESTIONS[29],  # code review philosophy
]
for q in sample_qs:
    print(f'\nQ: {q.text}')
    print(f'   principle: {q.principle.value}')
    top = _L18.match_stories(q, library, top_k=3)
    for story, score in top:
        print(f'   [{score}] {story.title}')
````

````python
# === Section 5: 3-2-1 rehearsal timings ===
rehearsals = _L18._build_rehearsals(library)
print(f'{"story":<48}  {"3-min":>5}  {"2-min":>5}  {"1-min":>5}  {"total":>5}')
print('-' * 80)
for title, r in rehearsals.items():
    t3 = len(r.three_minute.split()) / 150
    t2 = len(r.two_minute.split()) / 150
    t1 = len(r.one_minute.split()) / 150
    print(f'{title[:48]:<48}  {t3:>4.1f}m  {t2:>4.1f}m  {t1:>4.1f}m  {r.estimated_minutes():>4.1f}m')
````

````python
# === Section 6: Bank + principle coverage report ===
print('Bank coverage:')
for bank in _L18.STORY_BANKS:
    n = len(library.by_bank(bank))
    marker = 'OK ' if n >= 1 else 'GAP'
    print(f'  [{marker}] {bank:<10} {n} stories')
print('\nPrinciple coverage:')
for p in _L18.LeadershipPrinciple:
    n = len(library.by_principle(p.value))
    marker = 'OK ' if n >= 1 else 'GAP'
    print(f'  [{marker}] {p.value:<14} {n} stories')
````

````python
# === Section 7: A full STAR story on the wire ===
pick = _L18.SAMPLE_STORIES[0]  # the RAG migration
print(f'STORY: {pick.title}\n')
for label, field_name in [('Situation', 'situation'), ('Task', 'task'),
                          ('Action', 'action'), ('Result', 'result')]:
    print(f'{label}: {getattr(pick, field_name)}\n')

# Score it
sc = _L18.score_story(pick)
print(f'Heuristic score: {sc.total}/10')
for note in sc.notes:
    print(f'  note: {note}')

# Find the 3-2-1
r = rehearsals[pick.title]
print(f'\n3-min talk (first 400 chars):\n  {r.three_minute[:400]}...')
print(f'\n1-min Q&A:\n  {r.one_minute}')
````

## Architecture (Mermaid)

```mermaid
flowchart LR
    S[STARStory<br/>Situation / Task<br/>Action / Result] --> LIB[STARLibrary<br/>add / by_tag / by_role / by_bank]
    LIB --> P[principles_covered<br/>tag_for_principle]
    P --> PR[LeadershipPrinciple<br/>vision / execution /<br/>collab / tech / mentor]
    LIB --> M[match_stories<br/>question -> top-3]
    MOCK[MockQuestion<br/>30 common questions] --> M
    LIB --> SC[score_story<br/>quant / clear / impact / dur / role]
    S --> R[Rehearsal<br/>3-min / 2-min / 1-min]
    SC --> R
```

**Read as:** Stories go into the library; the library is indexed by tag, role, bank, and principle. Questions flow in; the matcher returns the top-3 stories. Each story has a 3-2-1 rehearsal and a heuristic score.
