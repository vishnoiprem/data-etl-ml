## Concept (5 min)\n\nThe modern AI app is **four layers**: (1) the foundation model (GPT-4o, Claude 3.5 Sonnet, Gemini, Llama 3), (2) the orchestration layer (LangChain, LlamaIndex, or raw `openai` calls), (3) the data layer (vector DB, SQL DB, cache), (4) the application layer (FastAPI, Next.js). Most failures happen because engineers pick the wrong layer. A question like 'summarize this PDF' needs vector retrieval, not a fine-tuned model. A question like 'extract line items from an invoice' needs structured output, not a 70B model. The cost-quality-latency trade-off is the central decision. GPT-4o is the highest quality but most expensive. GPT-4o-mini is 30x cheaper and 90% as good for simple tasks. Claude Haiku is fast and cheap. Gemini Flash is the cheapest with vision. Pick the model per call, not per app.\n
