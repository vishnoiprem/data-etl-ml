## Concept (5 min)\n\nThe Python AI toolkit in 2026 has stabilized around a handful of libraries. `openai` and `anthropic` are the official SDKs. `tiktoken` for token counting. `langchain` and `llama-index` for orchestration (but use sparingly -- raw API calls are often clearer). `pydantic` for structured outputs. `httpx` and `tenacity` for resilient HTTP. `rich` for terminal UIs. `pytest` and `pytest-asyncio` for tests. The dev environment is `uv` (fast pip replacement) + `pyenv` (Python versions) + `direnv` (per-dir env vars) + VS Code or Cursor. API keys are managed via `.env` files (gitignored) or a secrets manager in production. Never commit a key. Never log a key. The 80/20 of AI engineering setup: `uv venv`, `uv pip install openai anthropic pydantic tenacity httpx`, copy `.env.example` to `.env`, set `OPENAI_API_KEY=sk-...`.\n
