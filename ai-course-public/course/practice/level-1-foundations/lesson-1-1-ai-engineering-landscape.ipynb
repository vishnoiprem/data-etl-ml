## Concept (5 min)\n\nAI Engineering is the discipline of **building production systems on top of foundation models**. It is not training models. It is not academic research. It is shipping software that uses GPT-4, Claude, Gemini, or open-source models as a component. The four roles look similar but have very different day-to-day work. ML Engineers fine-tune and train. Data Scientists analyze and report. Prompt Engineers optimize prompts. AI Engineers architect end-to-end systems -- prompts, context windows, vector stores, tool calls, evals, observability, and cost controls. In 2026, the AI Engineer is the highest-leverage role in the stack because they multiply everyone else's output. The salary range is $150K-$300K+ for mid-to-senior, and the demand curve is still steep. The six core skills every AI engineer needs: (1) Prompt engineering, (2) RAG architecture, (3) Agent design, (4) Observability, (5) Cost optimization, (6) Production deployment. This course covers all six.\n
