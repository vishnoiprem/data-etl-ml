## Concept (5 min)\n\nThe OpenAI API has **three endpoints** you will use daily: `chat.completions.create`, `embeddings.create`, and `images.generate`. The chat completions endpoint accepts: model (gpt-4o, gpt-4o-mini, o1, o3-mini), messages (list of {role, content}), temperature (0-2), max_tokens, stream, tools (function calling), response_format (json_object or json_schema). Function calling is how the model returns structured arguments for a tool you defined. JSON mode forces valid JSON output. The `o1` and `o3` reasoning models are different: they think internally, have a separate `reasoning_tokens` field in usage, and cost more. **The architect decision** is which model per call. Use gpt-4o-mini for classification/extraction. Use gpt-4o for complex reasoning. Use o3-mini for math/logic. The cost difference is 30x. Build a wrapper that takes a model name and lets the rest of the app be model-agnostic.\n
