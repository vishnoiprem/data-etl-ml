## Concept (5 min)\n\nLangSmith is the **leading AI observability platform** from LangChain. Features: (1) **Tracing** -- automatic trace of every LLM call, including nested chains, retrievers, tools. UI shows the full tree. (2) **Datasets** -- upload test cases, run evals against any model, compare runs. (3) **Production monitoring** -- sample 1-100% of production traffic, run LLM-as-judge on each, alert on quality drops. (4) **Hub** -- share prompts and chains. (5) **Pricing** -- free tier (5K traces/mo), $39/mo (100K), enterprise. **Setup**: set `LANGCHAIN_TRACING_V2=true` and `LANGCHAIN_API_KEY=...` env vars. Then every LangChain call is automatically traced. For non-LangChain code, use the `@traceable` decorator or the OpenTelemetry integration.\n
