## Concept (5 min)\n\nLoad testing AI apps is **different from regular apps** because: (1) **Latency is high and variable** -- LLM calls take 1-30s with high variance. (2) **Rate limits are aggressive** -- OpenAI has per-minute token limits. (3) **Cost scales with load** -- 1M test requests could cost $500. (4) **Cold starts matter** -- the first request after a deploy is slow. **Tools**: **Locust** (Python, easy), **k6** (JS, modern, CI-friendly), **Artillery** (JS, scenarios). **The test**: ramp from 10 to 100 concurrent users, measure p50/p95/p99 latency, error rate, tokens/sec. **Capacity planning**: at 100 concurrent users, can your backend + the LLM API keep up? At what point do you start seeing rate limits? **Architect rule**: always load test before launch. Surprises at scale are bad.\n
