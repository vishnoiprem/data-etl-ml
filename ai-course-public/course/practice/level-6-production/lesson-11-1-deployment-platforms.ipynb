## Concept (5 min)\n\nChoosing a deployment platform: match it to your **scale and ops tolerance**. **PaaS (easiest)**: **Vercel** ($20/mo, Next.js-first, serverless), **Railway** ($5/mo+usage, simple, fast deploy), **Render** ($7/mo, similar to Railway), **Fly.io** ($0-30/mo, global edge, more control). **Cloud (more control)**: **AWS ECS** (Docker, complex, $50+/mo), **AWS Lambda** (serverless, $0-50/mo), **GCP Cloud Run** (serverless containers, $0-50/mo), **AWS SageMaker** (for ML, $100+/mo). **For AI apps**: Vercel + Railway for the prototype ($0-50/mo), Cloud Run or ECS for production scale ($100-500/mo), SageMaker or self-hosted GPUs for model serving ($500+/mo). **Architect decision**: start PaaS, move to cloud when you outgrow it (typically at $500-1000/mo PaaS bill or 100K+ requests/day).\n
