## Concept (5 min)\n\nOpenAI fine-tuning lets you customize `gpt-4o-mini` or `gpt-3.5-turbo` on your data. The process: (1) **Prepare data** -- JSONL format, each line `{'messages': [...]}`. Minimum 10 examples, recommended 100-1000. (2) **Upload** -- `client.files.create(file=..., purpose='fine-tune')`. (3) **Create job** -- `client.fine_tuning.jobs.create(training_file=..., model='gpt-4o-mini-2024-07-18')`. (4) **Wait** -- jobs take 10 min to 4 hours. (5) **Evaluate** -- run your test set on the new model. (6) **Deploy** -- use the fine-tuned model name in your API calls. **Cost**: training is $0.50-$50 depending on model and data size; inference is 4-8x the base model price. **Architect rule**: always keep a fallback to the base model -- if the fine-tuned model is down or underperforming, you can switch in seconds.\n
