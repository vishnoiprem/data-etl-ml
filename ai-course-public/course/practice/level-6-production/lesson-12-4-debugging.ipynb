## Concept (5 min)\n\nWhen vendor tools aren't enough, you build **your own observability stack**. The components: (1) **Structured logging** -- JSON logs with consistent fields (`timestamp, level, user_id, model, tokens, cost, latency, error`). Use `structlog` or `loguru`. (2) **Metrics** -- emit Prometheus metrics (`ai_requests_total`, `ai_tokens_total`, `ai_cost_dollars`). (3) **Traces** -- OpenTelemetry SDK, export to your backend. (4) **Custom dashboards** -- Grafana with your metrics. (5) **Alerting** -- PagerDuty/Opsgenie when metrics cross thresholds. **AI-specific alerts**: cost >$X/hour, error rate >Y%, latency p99 >Z, hallucination rate >W% (via eval sampling). **Architect rule**: every team should have a 'AI health' dashboard that shows the 5 key signals in one screen.\n
