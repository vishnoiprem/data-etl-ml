## Concept (5 min)\n\nCaching is the **single biggest cost saver** for AI apps. Three levels: (1) **L1: exact-match cache** -- hash the query, check Redis. Hit rate: 10-30% for general chatbots, 60-80% for FAQ-style apps. Sub-ms latency, 100% cost save on hits. (2) **L2: semantic cache** -- embed the query, check a vector store for similar queries above a threshold. Hit rate: 20-50%. ~10ms latency, 100% cost save on hits. (3) **L3: response cache** -- cache the full LLM response, not just the query. Useful for static prompts. **Invalidation**: TTL (time-based), event-based (when data changes), manual (admin). **Architect rule**: start with L1, add L2 if you have high query redundancy, add L3 for static content. The total cost saving is typically 50-80%.\n
