## Concept (5 min)\n\nCost optimization is a **continuous process**, not a one-time task. The 7 levers in order of impact: (1) **Model selection per call** -- 10-30x savings. (2) **Caching** -- 50-80% savings. (3) **Prompt compression** -- 20-40% savings. (4) **Batching** -- 30-50% savings at high volume. (5) **Model cascading** -- use small model first, large only if small is uncertain. 60-80% savings. (6) **Max output tokens** -- discipline on max_tokens. 10-20% savings. (7) **Reserved capacity** -- Azure/OpenAI committed-use discounts. 20-40% savings at scale. **Production case study**: a chatbot went from $50K/mo to $8K/mo by: switching to gpt-4o-mini (60% off), adding semantic cache (30% off), implementing model cascading (10% off). **Architect rule**: log every cost, review monthly, run a cost-optimization review quarterly.\n
