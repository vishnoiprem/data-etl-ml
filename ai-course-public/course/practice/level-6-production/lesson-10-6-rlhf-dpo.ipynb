## Concept (5 min)\n\n**RLHF** (Reinforcement Learning from Human Feedback) is the original technique that made ChatGPT work. The pipeline: (1) SFT (supervised fine-tuning) on demonstrations, (2) train a reward model on human preferences (A is better than B), (3) fine-tune the SFT model with PPO to maximize the reward. It's complex, unstable, and slow. **DPO** (Direct Preference Optimization) is a 2024 breakthrough that skips the reward model. It directly fine-tunes the model on preferences using a simple loss. Easier to implement, more stable, often better results. **When to use**: RLHF when you have a strong reward signal and need fine control. DPO for most cases. **Architect rule**: DPO is the default in 2026. Use RLHF only if DPO isn't working. **The catch**: you still need a high-quality preference dataset (1K-100K pairs).\n
