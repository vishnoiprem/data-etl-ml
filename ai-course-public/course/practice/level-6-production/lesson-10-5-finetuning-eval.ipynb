## Concept (5 min)\n\nFine-tuned model evaluation goes **beyond accuracy**. You need to check: (1) **Task quality** -- does it do the task well? (accuracy, F1, etc.) (2) **General quality** -- did you break general capabilities? (run a general benchmark like MMLU). (3) **Bias** -- does it perform differently across demographics? (test on sliced data). (4) **Safety** -- does it still refuse harmful requests? (run a red-team set). (5) **Latency/cost** -- is it slower or more expensive than the base model? (6) **A/B test in production** -- deploy to 10% of traffic, compare metrics for 1 week. **Architect rule**: never replace a model without an A/B test. Even if your offline eval is great, real users may behave differently.\n
