## Concept (5 min)\n\nHelicone is the **open-source alternative to LangSmith**. It's a proxy: you change your OpenAI base URL to `https://oai.helicone.ai/v1`, and every request is logged. Features: (1) **Dashboard** -- requests, tokens, cost, latency, by user/model. (2) **Caching** -- built-in semantic cache. (3) **Rate limiting** -- per-user. (4) **Self-hostable** -- Docker compose. (5) **Pricing** -- free for hobby, $20+/mo for teams. **OpenLLMetry** is the **OpenTelemetry instrumentation** for LLMs. It emits standard OTEL spans, so you can use any OTEL-compatible backend (Datadog, Honeycomb, Grafana, Jaeger). The combination: OpenLLMetry for traces, your favorite OTEL backend for storage. **Architect decision**: Helicone for fast setup, OpenLLMetry for vendor-neutral / multi-cloud.\n
