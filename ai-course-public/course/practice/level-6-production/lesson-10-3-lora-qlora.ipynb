## Concept (5 min)\n\nFull fine-tuning of a 70B model needs 16x A100s ($50+/hour) and 500GB of VRAM. **LoRA** (Low-Rank Adaptation) fine-tunes only ~0.1% of the parameters, fitting in 1x A100 (80GB) and taking 1-2 hours. **QLoRA** goes further: 4-bit quantization of the base model + LoRA adapters, fitting a 70B model in 1x A100 (48GB). Quality: LoRA is ~95-99% of full fine-tuning, QLoRA is ~93-97%. **Tools**: `peft` (Hugging Face), `bitsandbytes` (quantization), `trl` (training), `accelerate` (distributed). **The recipe**: (1) load model in 4-bit, (2) add LoRA adapters (rank 16, alpha 32), (3) train on your data, (4) merge LoRA weights back into the base model for inference, (5) save or push to HuggingFace Hub. **Architect rule**: always use QLoRA over full fine-tuning unless you have a specific reason. The cost savings are 10x.\n
