## Concept (5 min)\n\nStripe is the **default payment platform**. The integration: (1) **Create products** in Stripe Dashboard (or via API). (2) **Create prices** -- recurring for subscriptions, metered for usage. (3) **Checkout** -- redirect to Stripe-hosted checkout (easiest) or use Stripe Elements (more control). (4) **Webhooks** -- listen for `customer.subscription.created`, `customer.subscription.updated`, `customer.subscription.deleted`, `invoice.paid`, `invoice.payment_failed`. (5) **Sync to your DB** -- on webhook, update your `subscriptions` table. (6) **Gate features** -- check subscription status before allowing paid features. **Usage-based billing**: Stripe metered billing -- report usage at the end of each billing cycle via `usage_records.create`. **Architect rule**: never trust the frontend for payment status -- always check your DB (which was updated by the webhook).\n
