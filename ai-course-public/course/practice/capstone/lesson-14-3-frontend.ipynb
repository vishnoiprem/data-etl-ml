## Concept (5 min)\n\nThe frontend is the **only part the user sees**. It must be fast, beautiful, and forgiving. **Core components**: (1) **Input box** -- the only persistent UI. Autofocus, Enter to submit, Shift+Enter for newline. (2) **Message list** -- the conversation. User messages right-aligned, AI left-aligned. (3) **Streaming** -- show tokens as they arrive. Use SSE on backend, EventSource on frontend. (4) **Loading states** -- show a typing indicator (3 animated dots) while waiting. (5) **Error UI** -- if the AI fails, show a friendly message with a 'Retry' button. Never show a raw error. (6) **Markdown** -- render AI responses as markdown (use `react-markdown`). (7) **Code highlighting** -- for code blocks (use `react-syntax-highlighter`). **Architect rule**: the frontend is 30% of the work but 100% of the user's perception.\n
