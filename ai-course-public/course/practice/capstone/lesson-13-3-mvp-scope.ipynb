## Concept (5 min)\n\nAn **MVP is not a small product -- it's the smallest product that delivers value**. The discipline: if a feature doesn't directly serve the core value prop, it's out. The 2-week MVP plan: **Week 1**: backend (auth, DB, 1 core AI feature), basic frontend (1 page, 1 form). **Week 2**: polish, Stripe, deploy. That's it. **Feature prioritization frameworks**: (1) **RICE** -- Reach * Impact * Confidence / Effort. (2) **MoSCoW** -- Must, Should, Could, Won't. (3) **Kano** -- Basic (must), Performance (more is better), Excitement (wow factor). **Technical debt strategy**: ship fast, but write the parts that are hard to change later (DB schema, auth, payments) carefully. Skip the parts that are easy to refactor (UI, helper functions). **Architect rule**: if a feature takes >1 week, cut it. You can add it in v2.\n
