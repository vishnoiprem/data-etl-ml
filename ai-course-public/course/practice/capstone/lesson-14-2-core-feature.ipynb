## Concept (5 min)\n\nThe core AI feature is the **heart of your product**. Build it in 3 layers: (1) **Prompt layer** -- the system prompt, the user prompt template, the output schema. Versioned in your repo. (2) **Integration layer** -- the LLM call with retries, fallbacks, cost tracking. Wraps the prompt layer. (3) **API layer** -- the endpoint that the frontend calls. Validates input, calls integration, returns output. **Test with real users**: ship to 5 people you know, watch them use it, take notes. Don't ask them what they want -- watch what they do. **Iteration**: the v1 will be wrong. That's expected. Ship, get feedback, fix, repeat. **Architect rule**: the prompt is a config file, not code. Make it editable without a deploy.\n
