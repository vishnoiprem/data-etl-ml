## Concept (5 min)\n\nFew-shot prompting means including **examples in the prompt** to teach the model the pattern. Zero-shot (no examples) works for tasks the model already knows. One-shot (1 example) sets the format. Few-shot (3-10 examples) teaches nuance. The examples should be **diverse** (cover edge cases), **correct** (the model copies your mistakes), and **ordered** (hardest last, simplest first, or vice versa depending on the model). Counter-intuitive: **few-shot can hurt** when the examples are too similar -- the model overfits to the pattern and misses the general rule. When to use few-shot: classification, extraction, format-sensitive generation. When NOT: reasoning, math, creative writing. For reasoning, use chain-of-thought instead. Example selection strategies: random, k-nearest-neighbor to the test input, hard examples (where the model failed before).\n
