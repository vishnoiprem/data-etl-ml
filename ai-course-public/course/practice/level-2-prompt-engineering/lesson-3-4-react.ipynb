## Concept (5 min)\n\nReAct (Reasoning + Acting) is the pattern behind every modern agent. The loop: (1) **Thought**: the model thinks about what to do next. (2) **Action**: it picks a tool and arguments. (3) **Observation**: the tool result comes back. (4) **Repeat** until the model decides to output a final answer. The prompt format is strict: `Thought: ...\nAction: tool_name(arg1=value1)\nObservation: ...` The model is trained on this pattern in the system prompt. The **architect decisions** are: how many turns before forcing a final answer (default 10), what to do if the model gets stuck in a loop (detect repetition, force a different action), how to handle tool errors (retry once, then escalate). ReAct is the foundation for LangChain agents, OpenAI Assistants, and Claude tool use.\n
