## Concept (5 min)\n\nPrompt chaining is **breaking a complex task into smaller prompts**. A single 2000-word prompt asking the model to 'write a research report' produces mediocre output. A chain of 5 prompts (research, outline, draft sections, review, polish) produces excellent output because each step is focused. Three patterns: (1) **Sequential** -- output of step 1 is input to step 2. (2) **Parallel** -- independent steps run simultaneously, results merged. (3) **Conditional** -- different paths based on intermediate output. LangChain LCEL (LangChain Expression Language) is the de facto chaining syntax: `chain = prompt1 | model | parser | prompt2 | model | parser`. The **architect decision** is granularity. Too few steps = bad output. Too many = high latency, high cost. The sweet spot is 3-7 steps for a complex task.\n
