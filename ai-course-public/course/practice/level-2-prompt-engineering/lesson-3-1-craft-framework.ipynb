## Concept (5 min)\n\nMost prompts fail because they miss one of the five CRAFT elements. **Context**: what's the background, what does the model know? **Role**: who is the model acting as? **Action**: what should it do (verb)? **Format**: how should the output be structured (JSON, markdown, bullet list)? **Tone**: formal, casual, technical? A weak prompt: 'summarize this article.' A CRAFT prompt: 'You are a senior editor at The Economist (Role). Read the article below (Context). Produce a 5-bullet executive summary (Action). Output as a markdown list with bold headlines (Format). Use a formal, analytical tone (Tone).' The difference: the second prompt gets usable output 90% of the time; the first gets usable output 50% of the time. The framework is a checklist, not a template -- use the elements that matter for your task.\n
