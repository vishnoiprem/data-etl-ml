## Concept (5 min)\n\nLLMs have **no memory** -- every call is independent. To have a conversation, you re-send the full history. The challenge: history grows unbounded, context windows are finite, costs scale linearly. Four memory patterns: (1) **Full history** -- send every message. Works for <20 turns. (2) **Sliding window** -- keep last N turns. Simple, loses old context. (3) **Summarization** -- compress old turns into a summary. Saves tokens, loses detail. (4) **Vector memory** -- store turns in a vector DB, retrieve the K most relevant to the current query. Most powerful, most complex. The **architect decision** is budget: how many tokens can this conversation use? At 10K tokens, full history is fine. At 100K, you need summarization or vector memory. Production pattern: sliding window for recent (last 5 turns) + vector retrieval for relevant old (top 3).\n
