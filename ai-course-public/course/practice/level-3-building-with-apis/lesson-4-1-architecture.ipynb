## Concept (5 min)\n\nA production AI app has **three layers**: (1) **Frontend** (Next.js, Streamlit, native iOS) -- handles auth, UI, optimistic updates. (2) **Backend** (FastAPI, Express) -- handles business logic, queues long-running tasks, manages user state. (3) **AI layer** (separate package or service) -- wraps LLM calls, embeddings, vector search. Why three? Because AI calls are slow (1-30s) and the AI layer needs to scale independently. The **sync vs async decision**: if the user is waiting (chat, code completion), stream synchronously. If the task takes >5s and the user can leave (email generation, document analysis), enqueue it and email/webhook when done. **Queue systems**: Celery (Python), BullMQ (Node), Inngest (language-agnostic). The folder structure: `app/` (backend), `app/ai/` (AI calls), `app/workers/` (queued jobs), `frontend/` (Next.js).\n
