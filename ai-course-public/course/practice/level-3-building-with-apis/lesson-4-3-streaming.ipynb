## Concept (5 min)\n\nStreaming is **essential** for chat UX. Without it, users stare at a spinner for 5-10 seconds. With it, they see tokens appear in <500ms. Three streaming protocols: (1) **Server-Sent Events (SSE)** -- one-way, server to client, simple, works over HTTP. Default for AI. (2) **WebSockets** -- bidirectional, lower latency, more complex. Use for collaborative apps. (3) **Long polling** -- old-school, avoid. On the backend, the LLM SDK returns chunks when `stream=True`. You forward each chunk to the client via SSE. On the frontend, you append chunks to a buffer and update the DOM. **Architect concerns**: use a queue (Redis Streams, Kafka) for multi-instance deployments so a long stream survives a worker restart. Set a max stream duration (60s default) to avoid runaway costs. Always include a `done` event with usage stats.\n
