## Concept (5 min)\n\nMulti-modal apps combine **2+ modalities** (text, image, audio, video) in one interface. The challenge: each modality has different latency. Text: 200ms. Image: 1-3s. Audio generation: 1-5s. Video: 30s-5min. **UX pattern**: show what you can quickly, queue what you can't. Show the text response immediately, generate the audio in parallel. **Latency optimization**: (1) Pre-generate predictable content (greetings, error messages). (2) Stream everything possible. (3) Use smaller models for the first response, upgrade on follow-up. (4) Run modalities in parallel. **Architect concern**: the model choice. GPT-4o accepts all modalities natively. Otherwise, you need a transcription step (Whisper), a vision step (GPT-4V), etc. The unified API is simpler; the multi-model is cheaper.\n

## Build It: Spec\n\nBuild a multi-modal assistant: (1) accepts text OR image OR audio, (2) transcribes/images the input if needed, (3) generates a text response, (4) optionally speaks the response, (5) measures end-to-end latency. Mock all the AI calls. Demo: text query (fast), image query (slower), audio query (slowest).\n
