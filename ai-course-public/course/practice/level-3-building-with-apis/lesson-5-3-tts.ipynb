## Concept (5 min)\n\nTTS in 2026 is **indistinguishable from humans** for short clips. Options: **OpenAI TTS** ($15/$1M chars for HD, 6 voices), **ElevenLabs** ($5-$330/mo, best quality, voice cloning), **Play.ht** ($14-$99/mo), **Coqui XTTS** (open-source, free). **Voice cloning** raises ethical questions: always get consent, never clone without disclosure. **Streaming audio**: chunk the response and send via SSE/WebSocket.
