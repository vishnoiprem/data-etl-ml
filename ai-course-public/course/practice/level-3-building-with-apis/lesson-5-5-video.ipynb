## Concept (5 min)\n\nVideo generation is the **newest and most expensive** AI modality. Models: **Sora** (OpenAI, 1080p, up to 20s, $0.10-$0.50/second), **Runway Gen-3** ($0.05-$0.12/second), **Pika** ($0.05-$0.30/second), **Luma Dream Machine** ($0.10-$0.50/second), **Stable Video Diffusion** (open-source, free if self-hosted). **Current limits**: 5-20 second clips, 720p-1080p, no coherent narrative, physics is hit-or-miss, hands and text often wrong. **Use cases**: marketing clips, social media, prototyping, B-roll. **Architect decision**: 5-second social clip on Sora = $0.50. 1000 clips = $500. Self-hosting SVD on 1x A100 = $1.50/hour = can do 200 clips/hour. Break-even at ~3000 clips. **Don't use video generation** when: you need >30s, you need precise control, you need dialogue. Use real video.\n
