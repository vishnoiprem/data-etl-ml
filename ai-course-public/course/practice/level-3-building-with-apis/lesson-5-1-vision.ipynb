## Concept (5 min)\n\nVision models accept **images as input** and answer questions about them. The three major options: (1) **GPT-4o** -- best quality, $5/$15 per 1M tokens (input tokens count image tokens, ~85 tokens per 512x512 tile). (2) **Claude 3.5 Sonnet** -- strong on documents and charts, $3/$15. (3) **Gemini 1.5 Flash** -- cheapest, $0.075/$0.30, native multi-modal. Use cases: OCR (extract text from images), image captioning, UI understanding (screenshot -> code), chart analysis, document Q&A. **Architect insight**: image tokens are expensive. A 1024x1024 image is ~765 tokens. A 20-page PDF with images is ~10K tokens. Always resize before sending. Always use `detail: 'low'` (85 tokens flat) when you don't need fine detail. The cost difference between `detail: 'low'` and `detail: 'high'` is 9x.\n
