## Concept (5 min)\n\nDocument AI is **extracting structured data from messy documents** (PDFs, scanned images, contracts, invoices, receipts). The pipeline: (1) **Parse** -- PyMuPDF for text+layout, Unstructured for complex docs, Tesseract for OCR on scanned. (2) **Chunk** -- split into sections (headers, tables, paragraphs). (3) **Extract** -- send each chunk to an LLM with a Pydantic schema. (4) **Validate** -- check totals, dates, IDs against rules. **Tools**: PyMuPDF (fast, text + images), pdfplumber (best for tables), Unstructured (handles complex layouts), LlamaParse (LLM-based, $0.003/page). **Architect decision**: rule-based (regex, heuristics) for known formats, LLM-based for arbitrary docs. LLM is 10x slower but handles 10x more document types.\n
