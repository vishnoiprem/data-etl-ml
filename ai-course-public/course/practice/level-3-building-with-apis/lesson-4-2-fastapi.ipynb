## Concept (5 min)\n\nFastAPI is the default AI backend framework in 2026 because it's async-first, has built-in OpenAPI docs, and integrates with Pydantic. For AI apps, the key features are: (1) **Async endpoints** -- `async def` so the event loop can handle other requests while waiting for OpenAI. (2) **WebSocket** -- for streaming chat, bidirectional updates. (3) **Pydantic models** for request/response -- type safety and auto-validation. (4) **Dependency injection** for auth, rate limiting, DB sessions. (5) **Background tasks** for fire-and-forget jobs. The pattern for an LLM call: receive request -> validate -> call LLM (async) -> stream or return -> log usage. **Auth**: use Bearer tokens (JWT or API keys) in the `Authorization` header. **Rate limiting**: Redis-based token bucket, per-user, configurable RPM and TPM.\n
