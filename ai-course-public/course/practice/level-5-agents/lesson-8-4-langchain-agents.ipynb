## Concept (5 min)\n\nLangChain has **4 agent types**: (1) **ReAct** -- text-based reasoning, works with any LLM. (2) **OpenAI Functions** -- uses function calling, more reliable with OpenAI models. (3) **Plan-and-Execute** -- planner creates a plan upfront, executor follows it. Better for long tasks. (4) **Multi-Action** -- can take multiple actions per turn. The **AgentExecutor** runs the agent loop with error handling, early stopping, and iteration limits. **Custom tools**: subclass `BaseTool` or use the `@tool` decorator. The tool needs a name, description (used by the LLM to decide when to use it), and a `_run` method. **Architect rule**: keep tools small and focused. 5 small tools > 1 big tool. Each tool should be <100 tokens of description.\n
