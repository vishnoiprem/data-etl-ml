## Concept (5 min)\n\nMulti-agent systems use **multiple specialized agents** that collaborate. When to use: (1) **Specialization** -- different agents for research, writing, code, review. Each is better at its narrow task. (2) **Parallelism** -- multiple agents work on different parts of a problem simultaneously. (3) **Separation of concerns** -- one agent doesn't need to know everything. (4) **Error recovery** -- if one agent fails, another can pick up. **When NOT**: a single agent with good prompts can do the task. Multi-agent adds 3-10x cost and 2-5x latency. **Real-world examples**: CrewAI (role-based crews), AutoGen (conversational), LangGraph (stateful graphs), Anthropic's multi-agent research system, ChatGPT's multi-agent team. **Architect decision**: start single-agent, add multi-agent only when you hit a complexity wall.\n
