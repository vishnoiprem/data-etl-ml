## Concept (5 min)\n\nThe OpenAI Assistants API is a **managed agent platform**. You create an Assistant (with instructions, model, tools), then Threads (one per conversation), then Runs (executions of the assistant on a thread). Built-in tools: **Code Interpreter** (Python sandbox, can run code, generate files), **File Search** (RAG over uploaded files, automatic chunking and retrieval), **Function Calling** (your custom tools). **Architect decision**: Assistants API vs raw chat.completions? Assistants is great for prototyping and for apps that need code execution. For high-throughput production, raw API + your own orchestration is faster and cheaper. Assistants has a $0.03/GB/day storage cost for files. **State management**: threads persist on OpenAI's side, so you don't have to manage conversation history yourself.\n
