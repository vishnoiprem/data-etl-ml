## Concept (5 min)\n\nMulti-agent systems need **communication protocols**. Patterns: (1) **Message passing** -- agents send messages to each other (Autogen style). Simple, but can lead to loops. (2) **Shared state** -- all agents read/write a shared state object (LangGraph style). Powerful, requires schema discipline. (3) **Handoffs** -- one agent hands off the task to another (OpenAI Swarm style). Clean, but linear. (4) **Blackboard** -- agents write to a shared 'blackboard', one reads and decides. (5) **A2A (Agent-to-Agent) protocol** -- emerging standard, JSON-RPC over HTTP. **Architect decision**: pick the simplest pattern that works. For 2-3 agents, message passing. For 3-10 agents with shared context, shared state. For 10+ agents, A2A or a dedicated orchestration service.\n
