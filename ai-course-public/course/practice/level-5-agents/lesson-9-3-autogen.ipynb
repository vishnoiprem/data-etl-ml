## Concept (5 min)\n\nAutoGen is Microsoft's multi-agent framework. The core idea: **agents converse** to solve a task. Each agent is a `ConversableAgent` with a `reply_func` (the LLM call). The conversation continues until a termination condition (max messages, a specific word, a function call). Patterns: (1) **Two-agent** -- UserProxyAgent + AssistantAgent, classic code execution. (2) **Group chat** -- multiple agents + a GroupChatManager that picks the next speaker. (3) **Nested chats** -- agents can spawn sub-conversations. **Human-in-the-loop**: UserProxyAgent can request human input via console or a callback. **Architect decision**: AutoGen is more flexible than CrewAI but more code. Use for: research simulations, code teams, anything that needs back-and-forth. Don't use for: simple sequential pipelines.\n
