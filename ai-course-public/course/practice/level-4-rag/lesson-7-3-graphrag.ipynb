## Concept (5 min)\n\nGraphRAG is **RAG over a knowledge graph** instead of (or in addition to) a vector store. The advantage: it handles **multi-hop questions** that pure vector search can't ('Who are the investors in companies that acquired startups founded by people who went to MIT?'). The pipeline: (1) Extract entities and relationships from docs (using an LLM). (2) Store in a graph DB (Neo4j is the default). (3) At query time, traverse the graph to find relevant subgraphs. (4) Convert to text, feed to LLM. **Microsoft GraphRAG** is the open-source reference implementation. **Architect concern**: graph extraction is expensive (~$0.10 per page of docs). Graph storage is fast for traversal but slow for fuzzy/keyword search. **Use GraphRAG when**: your data is highly relational (people-companies-events), you need multi-hop reasoning, or vector search quality plateaus.\n
