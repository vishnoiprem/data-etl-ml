## Concept (5 min)\n\nRAG quality is **invisible without metrics**. The four core RAG metrics: (1) **Context precision** -- of the top-K retrieved docs, what fraction are relevant? (2) **Context recall** -- of the docs needed to answer, what fraction were retrieved? (3) **Faithfulness** -- is the answer grounded in the retrieved docs (no hallucination)? (4) **Answer relevance** -- is the answer relevant to the question? The standard framework: **RAGAS** (open-source, Python, $0 LLM-judge cost if you bring your own model). **Test set construction**: 50-200 question/answer/gold-docs triples, ideally from real user queries (with permission). **Architect rule**: run RAGAS on every prompt change, every embedding model change, every chunk size change. Treat it as a CI check.\n
