## Concept (5 min)\n\nChunking is the **most underrated part of RAG**. A 100-page PDF won't fit in the context window. You split it into chunks of 500-2000 tokens, embed each, store them. At query time, you retrieve the top-K chunks. The chunking strategy determines retrieval quality. **Four strategies**: (1) **Fixed-size** (every 500 tokens) -- simple, breaks sentences. (2) **Recursive character** (split on \n\n, then \n, then space) -- respects paragraphs. (3) **Semantic** (split when embedding distance is high) -- best quality, slowest. (4) **Document-aware** (respect headers, code blocks, tables) -- best for structured docs. **Overlap** (10-20% of chunk size) prevents losing context at boundaries. **Metadata** (source, page, section) goes in the chunk -- you need it for citations. **Architect decision**: chunk size. Smaller (200-500) = better precision, more chunks, more cost. Larger (1500-3000) = more context per chunk, may dilute the signal. The sweet spot is 500-1000.\n
