## Concept (5 min)\n\nRAG for code is **different from RAG for prose**. Code has structure: functions, classes, imports, comments. Naive character splitting breaks function definitions in the middle, which destroys retrieval quality. **Three code-aware chunking strategies**: (1) **AST-based** -- parse with `tree-sitter` (Python, JS, Go, Rust, ...), split at function/class boundaries. Best quality. (2) **Language-specific** -- use the language parser (Python's `ast` module) to extract functions/classes. (3) **Sliding window with overlap** -- fallback when AST isn't available. **Embeddings**: code-aware models like `voyage-code-2` ($0.12/1M tokens) or `text-embedding-3-small` work for natural language queries over code. **Use cases**: codebase Q&A, onboarding, code review, docs generation. **Architect note**: always include file path, function name, and line numbers as metadata -- users will ask 'where is X defined?' and you need to point to the file.\n
