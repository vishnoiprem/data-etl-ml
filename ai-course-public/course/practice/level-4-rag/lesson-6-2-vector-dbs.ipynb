## Concept (5 min)\n\nA vector database stores **embeddings and supports fast nearest-neighbor search**. When you have 1M documents, you can't compare a query to all 1M at query time (too slow). Vector DBs use **approximate nearest neighbor (ANN)** algorithms: **HNSW** (Hierarchical Navigable Small World, fast, memory-hungry, default for most apps), **IVF** (Inverted File Index, fast on disk, less accurate), **ScaNN** (Google's, very fast on TPUs). The five main options: **Pinecone** ($0.096/1M vectors/mo, managed, easy), **Weaviate** (open-source, self-hosted or managed, flexible), **Qdrant** (open-source, Rust, fast), **Chroma** (open-source, Python-native, good for dev), **pgvector** (Postgres extension, use if you already have Postgres). **Decision rule**: pgvector for <1M vectors and you need joins, Pinecone for >10M vectors and zero ops, Qdrant for open-source + performance, Chroma for prototyping.\n
