## Concept (5 min)\n\nLlamaIndex is the **RAG-first framework**. It was built specifically for retrieval-augmented generation, while LangChain was built for general LLM chains. LlamaIndex shines for: (1) **Complex indexing** -- tree indexes, list indexes, keyword indexes, composable. (2) **Query engines** -- sub-question query engine, multi-document synthesis. (3) **Response synthesizers** -- tree summarize, compact, refine. (4) **Agents over data** -- ReAct agent with RAG tools. **When to pick LlamaIndex over LangChain**: you have a document-heavy use case, you need to index the same docs in multiple ways, or you want a higher-level abstraction. **When to pick LangChain**: you have many non-RAG LLM calls (agents, chains), you want more integrations. The code is shorter in LlamaIndex for RAG-only apps.\n
