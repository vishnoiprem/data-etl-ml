## Concept (5 min)\n\nAgentic RAG is **RAG where an agent decides when and how to retrieve**. The flow: user asks a question, the agent thinks: do I need to look anything up? (1) **No** (general knowledge) -> answer directly. (2) **Yes, simple lookup** -> vector search. (3) **Yes, multi-step** -> decompose, retrieve for each sub-question, synthesize. (4) **Yes, with filters** -> use metadata filters. (5) **Yes, fresh data** -> web search instead. The agent uses **tools**: vector_search, web_search, sql_query, calculator. The **ReAct loop** drives it. **Architect decision**: do you need agentic RAG? It adds 20-50% latency (more LLM calls) and 2-3x cost. Worth it for complex Q&A over heterogeneous sources. Not worth it for simple single-doc search.\n
