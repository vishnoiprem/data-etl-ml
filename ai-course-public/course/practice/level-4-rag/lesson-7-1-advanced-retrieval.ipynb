## Concept (5 min)\n\nBasic vector search leaves 30-40% of relevant docs on the table. **Advanced retrieval** combines multiple techniques: (1) **Hybrid search** -- BM25 (keyword) + vector (semantic), merged with reciprocal rank fusion. Catches exact-match queries vectors miss. (2) **Re-ranking** -- first retrieve 50, then re-rank with a cross-encoder (Cohere Rerank 3, or a local cross-encoder) to get the top 5. 20-30% quality boost. (3) **Query expansion** -- ask the LLM to generate 3-5 paraphrases of the query, retrieve with each, merge. (4) **HyDE** (Hypothetical Document Embeddings) -- generate a fake answer, embed it, retrieve with that. (5) **Multi-query** -- decompose the question into sub-questions, retrieve for each, combine. **Architect rule**: start with vector search, add hybrid if you see keyword misses, add re-ranking if you see ranking errors. Don't add all 5 upfront -- they're expensive.\n
