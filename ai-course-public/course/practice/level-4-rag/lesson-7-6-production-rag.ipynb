## Concept (5 min)\n\nProduction RAG is **engineering, not ML**. The patterns: (1) **Caching** -- exact-match (Redis) for repeated queries, semantic cache (GPTCache) for paraphrases. Saves 50-80% of LLM cost. (2) **Streaming** -- stream the LLM response while you retrieve. Show the user something in <500ms. (3) **Async** -- retrieve and embed in parallel; don't block the request thread. (4) **Error handling** -- if vector DB is down, return a graceful 'I can't search right now' instead of 500. (5) **Fallbacks** -- if retrieval returns nothing, generate from general knowledge (with a caveat in the response). (6) **Cost optimization** -- small model for grading/re-ranking, large model for generation. (7) **Multi-tenant isolation** -- filter all queries by tenant_id at the DB level. (8) **Observability** -- log every retrieval, every score, every token. You'll need it for debugging.\n
