## Concept (5 min)\n\nLangChain is the most popular RAG framework in 2026, mostly because of **LCEL** (LangChain Expression Language) -- a pipe-based syntax for chaining. A LCEL RAG chain: `retriever | format_docs | prompt | model | parser`. The retriever can be a vector store, BM25, or hybrid. LangChain has integrations for 100+ loaders (PDF, Notion, GitHub, S3), 30+ splitters, 20+ vector stores, 10+ models. The cost of LangChain: a complex abstraction layer that can be hard to debug. The benefit: 50 lines of code instead of 500. **Architect decision**: use LangChain if you need many integrations and your team knows it. Use raw openai + numpy if you want simplicity and full control. **Production pattern**: wrap LangChain in a thin internal interface so you can swap it out later.\n
