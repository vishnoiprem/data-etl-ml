## Concept (5 min)

Every AI agent deployment has **6 numbers**: 5 inputs (how much time, how much it's worth, how many tasks, how often it's used, how often it's right) and 1 cost (how much the LLM + infra costs per month). The **4 use case criteria** (repetitive, language-heavy, tool-using, measurable) tell you **which problems are even worth estimating**. The formula then tells you **what the answer is in dollars per month**.

The wrong choice is to skip the formula and just build. The right choice is to estimate the 6 numbers first, run them through `monthly_roi()`, and only then open your editor.

**The formula in code (this is the whole course in 2 lines):**

```python
monthly_value = hours_saved * rate * 4.33 * (volume/100) * adoption * success
monthly_roi   = monthly_value - agent_cost
```

## Architect Notes

### Trade-offs

| Choice | Pros | Cons | Pick when |
|---|---|---|---|
| 1 trial per use case | Fast | One bad draw can mislead you | Quick gut check |
| **1,000 trials (this course)** | **See the distribution; P05 matters** | 30 sec compute | **You deploy with real money** |
| 10,000 trials | Even tighter P05 | 5 min compute | Pre-IPO due diligence |
| Real measurement data | Ground truth | You don't have it yet | 3+ months post-launch |

### Capacity Model

The 1000-cycle review uses these ranges (sampled uniformly):

| Input | Range | Why |
|---|---|---|
| `hours_saved_per_week` | 1.0 - 5.0 | A solo SMB owner can't free more than ~5 hr/wk from one task |
| `hourly_rate_usd` | $25 - $150 | SGD 35 - 200 / hr equivalent |
| `volume_factor` | 10 - 100 | Weekly task count, not daily |
| `adoption_rate` | 50% - 100% | First month is 50%, mature agents hit 100% |
| `success_rate` | 70% - 100% | Below 70% the user abandons the agent |
| `agent_cost_usd` | $10 - $100/mo | LLM + observability + retries |

### When NOT to use this lab's content

- **Don't** use this formula for a 1-shot creative task -- it fails the `repetitive` criterion.
- **Don't** use it for a task where success is below 70% -- the math says positive ROI but the user churns.
- **Don't** use it when the hourly rate is "free" (a hobby) -- the rate is opportunity cost, not wages.
- **Do** use it for any task that hits all 4 criteria and you can put a $/hr on your time.
