```python
def longest_monotonic_run(arr: list[int]) -> int:
    if not arr:
        return 0

    max_len = 1
    inc = 1
    dec = 1

    for i in range(1, len(arr)):
        # Non-decreasing check
        if arr[i] >= arr[i - 1]:
            inc += 1
        else:
            inc = 1

        # Non-increasing check
        if arr[i] <= arr[i - 1]:
            dec += 1
        else:
            dec = 1

        max_len = max(max_len, inc, dec)

    return max_len
```
