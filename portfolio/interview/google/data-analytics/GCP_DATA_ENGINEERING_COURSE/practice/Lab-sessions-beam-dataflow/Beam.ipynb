## `beam pipeline creation : 2 methods`



```python
# first method:

with beam.Pipeline(options=beam_options) as pipeline:
    output = (
        pipeline
        | "read from source" >> beam.io.Read("path")  # pc1
        | "do transformation-1" >> beam.Map(function)  #pc2
        | "do transformation-2" >> beam.Filter(function)  #pc3
        | "do transformation-3" >> beam.ParDo(Class)  #pc4
        | "do transformation-4" >> beam.GroupByKey()  #pc5
        | "do transformation-5" >> beam.CombinePerKey(agg)  #pc6
        | "write to sink" >> beam.io.Write("path")
    )
```

```python
# second method:

with beam.Pipeline(options=beam_options) as pipeline:
    pc1 = pipeline | "Read from source" >> beam.io.Read("path")  
    pc2 = pc1 | "Do transformation-1" >> beam.Map(function)    
    pc3 = pc2 | "Do transformation-2" >> beam.Filter(function)    
    pc4 = pc3 | "Do transformation-3" >> beam.ParDo(Class)    
    pc5 = pc4 | "Do transformation-4" >> beam.GroupByKey()    
    pc6 = pc5 | "Do transformation-5" >> beam.CombinePerKey(agg)    
    pc6 | "Write to sink" >> beam.io.Write("path")
```

### `read from employee.csv file and perform below`

    - adding columns 
    - verify the schema
    - null => default values 
    - filtering
    - drop duplicates

```python
csv_path = "gs://src-bkt-sdk-13022025/employees.csv"

def transform_data(element):
    fields = element.split(",")
    
    if len(fields) < 6:
        return None
    
    try:
        name = fields[0].strip() or None
        age = fields[1].strip() or None
        gender = fields[2].strip() or None
        department = fields[3].strip() or None
        salary = fields[4].strip() or "0"
        country = fields[5].strip() or None
        origin = "India" # constant column
        salary_int = int(salary) if salary.isdigit() else 0
        tax = 0.1 * salary_int # derived column
        
    except Exception as e:
        print(f"Error processing record: {element} - {e}")
        return None 
    
    return (name, age, gender, department, salary, country, origin, tax)

with beam.Pipeline() as pipeline:
    
    output = (
        pipeline
        | "read from csv" >> beam.io.ReadFromText(csv_path, skip_header_lines=1)
        | "transform data" >> beam.Map(transform_data)
        | "filter the nulls" >> beam.Filter(lambda element: element is not None and element[0] != None)
        | "drop duplicates" >> beam.Distinct()
        | "Print output" >> beam.Map(print)
    )
```

## Aggregations

```python
csv_path = "gs://src-bkt-sdk-13022025/employees.csv" 

def transform_data(element):
    fields = element.split(",")
    
    if len(fields) < 6:
        return None
    
    try:
        name = fields[0].strip() or None
        age = fields[1].strip() or None
        gender = fields[2].strip() or None
        department = fields[3].strip() or None
        salary = fields[4].strip() or "0"
        country = fields[5].strip() or None
        origin = "India" # constant column
        salary_int = int(salary) if salary.isdigit() else 0
        tax = 0.1 * salary_int # derived column
        
    except Exception as e:
        print(f"Error processing record: {element} - {e}")
        return None 
    
    return (name, age, gender, department, salary, country, origin, tax)

with beam.Pipeline() as pipeline:
    input_data = (
        pipeline
        | "read from csv" >> beam.io.ReadFromText(csv_path, skip_header_lines=1)
        | "transform data" >> beam.Map(transform_data)
        | "filter the nulls" >> beam.Filter(lambda element: element is not None and element[0] != None)
        | "drop duplicates" >> beam.Distinct()
    )
    
    # total salary per department
    total_salary = (
        input_data
        | "extract dep, salary" >> beam.Map(lambda element: (element[3], int(element[4])))
        | "total salary" >> beam.CombinePerKey(sum)
        | "Print total salary" >> beam.Map(print)
    )
    
    # count of employees per department
    count_employees = (
        input_data
        | "extract dep, 1" >> beam.Map(lambda element: (element[3],1))
        | "total count" >> beam.CombinePerKey(sum)
        | "Print total emps" >> beam.Map(print)
    )
    
    # avg salary per department
    avg_salary = (
        input_data
        | "extract the dept, salary" >> beam.Map(lambda element: (element[3], (int(element[4]), 1))) 
        | "sum of salary, count" >> beam.CombinePerKey(lambda element: (sum(i[0] for i in element), sum(i[1] for i in element)))
        | "calculate avg salary" >> beam.Map(lambda x: (x[0], x[1][0]/x[1][1] if x[1][1] > 0 else 0))
        | "Print avg salary" >> beam.Map(print)
    )
    
```

## joins

```python
BUCKET_NAME = "src-bkt-sdk-13022025"

users_path = f"gs://{BUCKET_NAME}/users.txt"
orders_path = f"gs://{BUCKET_NAME}/orders.txt"

def parse_users(element):
    fields = element.split(",")
    if len(fields) != 3:
        return None
    user_id, name, country = fields[0], fields[1], fields[2]
    return (user_id, (name,country))

def parse_orders(element):
    fields = element.split(",")
    if len(fields) != 3:
        return None
    order_id, user_id, amount = fields[0], fields[1], fields[2]
    return (user_id, (order_id,amount))

def format_result(e):
    user_id, user_info, orders = e
    
    if user_info:
        name, country = user_info[0]
        
        for order in orders:
            order_id, amount = order
            return {"user_id":user_id,"name":name,"country":country,"order_id":order_id, "amount":amount}
    return None   


with beam.Pipeline() as pipeline:
    
    users = (
        pipeline
        | "Read users" >> beam.io.ReadFromText(users_path, skip_header_lines=1)
        | "parse users" >> beam.Map(parse_users)
        | "filter users" >> beam.Filter(lambda e: e is not None)
        # | "print users" >> beam.Map(print)
    )
    
    orders = (
        pipeline
        | "Read orders" >> beam.io.ReadFromText(orders_path, skip_header_lines=1)
        | "parse orders" >> beam.Map(parse_orders)
        | "filter orders" >> beam.Filter(lambda e: e is not None)
        # | "print orders" >> beam.Map(print)
    )
    
    # perform joins
    
    joined = (
        {"users" : users, "orders" : orders}
        | "join users and orders" >> beam.CoGroupByKey()
        | "format the results" >> beam.Map(lambda e: (e[0],e[1]["users"], e[1]["orders"]))
        | "format result" >> beam.Map(format_result)
        | "print joins" >> beam.Map(print)
    )
    
```

## clean and transform the user_events.csv with pardo

```python
from datetime import datetime

user_events_path = f"gs://{BUCKET_NAME}/user_events.csv"

class cleanAndTrasform(beam.DoFn):
    def process(self, element):
        fields = element.split(",")
        user_id = fields[0]
        event_type = fields[1]
        timestamp = fields[2]
        metadata = fields[3]+fields[4]
        
        # validate the user (should be numberic)
        if not user_id.isdigit():
            return None
        
        try:
            parsed_timestamp = datetime.strptime(timestamp, "%Y-%m-%dT%H:%M:%SZ")
            timestamp = parsed_timestamp.strftime("%Y-%m-%d %H:%M:%S")
        except ValueError:
            return None
        
        yield {
            "user_id" : user_id,
            "event_type" : event_type,
            "timestamp" : timestamp,
            "metadata" : metadata
        }        
        

with beam.Pipeline() as pipeline:
    output = (
        pipeline
        | "read events" >> beam.io.ReadFromText(user_events_path, skip_header_lines=1)
        | "clean and transform" >> beam.ParDo(cleanAndTrasform())
        | beam.Map(print)
    )
```
