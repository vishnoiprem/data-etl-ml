```python
from datetime import datetime, time ,timedelta

import json

result = {}

def sf_trip_months(trips: list[dict]) -> dict[str, str]:
    # print(trips)

    for row in trips:

        if row['status'] == 'completed' and  row['city']=='San Francisco':
            completed_utc= row['completed_utc']
            date_time = datetime.fromisoformat(completed_utc.replace("Z", "")) - timedelta(hours=8)
            
            yyyy_mm= date_time.strftime('%Y-%m')

            # result.append([row['trip_id'],yyyy_mm])
            result[row['trip_id']]=yyyy_mm

    return result 



trips=[
  {
    "city": "San Francisco",
    "status": "completed",
    "trip_id": "t1",
    "completed_utc": "2024-01-01T07:30:00Z"
  },
  {
    "city": "San Francisco",
    "status": "completed",
    "trip_id": "t2",
    "completed_utc": "2024-01-15T20:00:00Z"
  },
  {
    "city": "San Francisco",
    "status": "cancelled",
    "trip_id": "t3",
    "completed_utc": "2024-01-15T20:00:00Z"
  },
  {
    "city": "Los Angeles",
    "status": "completed",
    "trip_id": "t4",
    "completed_utc": "2024-01-15T20:00:00Z"
  }
]


print(sf_trip_months(trips))
```
