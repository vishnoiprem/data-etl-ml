```python
from typing import List

def cumulative_product(nums: List[int]) -> List[int]:
    # edge: empty
    if not nums:
        return []

    result = []
    prod = 1
    for x in nums:
        prod *= x # handles 0, negatives automatically
        result.append(prod)
    return result

# quick test
print(cumulative_product([2,3,4])) # [2,6,24]
print(cumulative_product([])) # []
print(cumulative_product([5])) # [5]
print(cumulative_product([2,0,4])) # [2,0,0]
```

```python
from collections import OrderedDict
from typing import Optional

def first_non_repeated_stream(s: str) -> str:
    freq = {}
    candidates = OrderedDict() # key -> True, order = first seen order

    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

        if freq[ch] == 1:
            # first time we see it — add to queue
            candidates[ch] = True
        elif freq[ch] == 2:
            # second time — it is now repeated, remove from queue if present
            candidates.pop(ch, None)
        # if freq > 2, already removed, do nothing

    # first key in OrderedDict is answer
    return next(iter(candidates), "")

print(first_non_repeated_stream("swiss")) # w
print(first_non_repeated_stream("aabccbdeff")) # d
```

```python
from typing import List

def diagonal_sum(matrix: List[List[int]]) -> int:
    if not matrix or not matrix[0]:
        return 0

    n = len(matrix) # n x n per question
    total = 0

    for i in range(n):
        total += matrix[i][i] # main: top-left -> bottom-right
        total += matrix[i][n - 1 - i] # anti: top-right -> bottom-left

    # odd n -> center counted twice, remove once
    if n % 2 == 1:
        mid = n // 2
        total -= matrix[mid][mid]

    return total

# tests
print(diagonal_sum([[1,2,3],[4,5,6],[7,8,9]])) # 25
print(diagonal_sum([[1,1,1,1],[1,1,1,1]])) # 8
print(diagonal_sum([[1,2],[3,4]])) # 10
print(diagonal_sum([[5]])) # 5 (center not doubled)
```

```python
from typing import List, Dict

def top_earning_titles(workers: List[Dict], titles: List[Dict]) -> List[str]:
    if not workers:
        return []

    # 1. Find max in O(n) — no sort
    max_salary = max(w["salary"] for w in workers)

    # 2. Top earner ids in O(n) — set for O(1) lookup
    top_ids = {w["worker_id"] for w in workers if w["salary"] == max_salary}

    # 3. Walk titles in input order, dedupe titles with set
    seen_titles = set()
    result = []
    for t in titles:
        wid = t.get("worker_id")
        ttitle = t.get("title")
        if wid in top_ids and ttitle not in seen_titles:
            seen_titles.add(ttitle)
            result.append(ttitle)

    return result

# ---- tests, run instantly on CoderPad ----
workers = [
 {"worker_id": 1, "salary": 100},
 {"worker_id": 2, "salary": 200},
 {"worker_id": 3, "salary": 200},
]
titles = [
 {"worker_id": 2, "title": "SDE"},
 {"worker_id": 3, "title": "SDE"}, # duplicate, keep first only
 {"worker_id": 2, "title": "Tech Lead"},
 {"worker_id": 3, "title": "Manager"},
 {"worker_id": 1, "title": "Intern"},
]

print(top_earning_titles(workers, titles)) # ["SDE", "Tech Lead", "Manager"]
print(top_earning_titles([], titles)) # []
print(top_earning_titles(workers, [])) # []
```

```python
from typing import List, Any

def flatten(feed: List[Any]) -> List[Any]:
    result = []

    def dfs(items):
        for x in items:
            if isinstance(x, list):
                dfs(x) # recurse into sub-list
            else:
                result.append(x) # leaf — int, str, etc stays whole

    dfs(feed)
    return result

# tests
print(flatten(feed))
print(flatten([[1,2],[3,[4,5]],6])) # [1,2,3,4,5,6]
print(flatten(["top",["trending",["sports","tech"]],"latest"])) # ["top","trending","sports","tech","latest"]
print(flatten([[], [[[]]], [1]])) # [1] — empty lists contribute nothing
print(flatten([])) # []
```

```python
def flatten_iter(feed: List[Any]) -> List[Any]:
    result = []
    stack = [feed] # start with whole feed

    while stack:
        curr = stack.pop()
        if isinstance(curr, list):
            # push reversed to keep left-to-right order
            for item in reversed(curr):
                stack.append(item)
        else:
            result.append(curr)

    return result
```

```python
def filter_odd_digits(s: str) -> str:
    # keep only '1','3','5','7','9' — no int() conversion, faster
    odd = {'1','3','5','7','9'}
    result = []

    for c in s:
        if c in odd: # O(1) set lookup, ignores letters safely
            result.append(c)

    return ''.join(result)

# or 1-liner Pythonic:
def filter_odd_digits(s: str) -> str:
    return ''.join(c for c in s if c in '13579')

print(filter_odd_digits("2468"))      # ''
print(filter_odd_digits("12345"))     # '135'
print(filter_odd_digits("a1b2c3"))    # '13' - doesn't crash
print(filter_odd_digits(""))          # ''
```

```python
from typing import List

def row_sums(matrix: List[List[int]]) -> List[int]:
    if not matrix:
        return [] # not 0

    result = []
    for row in matrix:
        total = 0
        for val in row:
            total += val
        result.append(total)
    return result

# Pythonic Staff version for interview:
def row_sums(matrix: List[List[int]]) -> List[int]:
    return [sum(row) for row in matrix]

print(row_sums([[1,2,3],[4,5,6],[7,8,9]])) # [6,15,24]
print(row_sums([[10,20],[30,40]])) # [30,70]
print(row_sums([])) # [] - not 0
print(row_sums([[], [1,2]])) # [0,3]
```

```python
def max_payoff(payoffs: list) -> int:
    if not payoffs:
        return 0

    prev2 = 0 # dp[i-2]
    prev1 = 0 # dp[i-1]

    for x in payoffs:
        if prev1 > prev2 + x:
             cur = prev1
        else:
            cur=prev2 + x

        prev2 = prev1
        prev1 = cur

    return prev1

print(max_payoff([2,7,9,3,1])) # 12 -> 2+9+1
print(max_payoff([2,1,1,2])) # 4 -> 2+2
print(max_payoff([])) # 0
print(max_payoff([5])) # 5
```

```python
def best_trade(prices: list[int]) -> list[int]:
    if len(prices) < 2:
        return [0, -1, -1]

    min_price = prices[0]
    min_idx = 0

    best_profit = 0
    best_buy = -1
    best_sell = -1

    for sell in range(1, len(prices)):
        profit = prices[sell] - min_price

        # better profit?
        if profit > best_profit:
            best_profit = profit
            best_buy = min_idx
            best_sell = sell
        # tie -> earliest buy, then earliest sell
        elif profit == best_profit and profit > 0:
            if min_idx < best_buy or (min_idx == best_buy and sell < best_sell):
                best_buy = min_idx
                best_sell = sell

        # update min for future - keep earliest index on equal price
        if prices[sell] < min_price:
            min_price = prices[sell]
            min_idx = sell

    return [best_profit, best_buy, best_sell]

print(best_trade([7,1,5,3,6,4])) # [5,1,4] 1->6
print(best_trade([7,6,4,3,1])) # [0,-1,-1] no gain
print(best_trade([1,2,3,1,2,3])) # [2,0,2] tie -> earliest
```

```python
def quantile_calculator(data: list[int | float], percentile: float) -> float:
    if not data:
        return 0

    a = sorted(data)
    n = len(a)

    if n == 1:
        return float(a[0])

    # clamp 0-100
    p = max(0, min(100, percentile)) / 100.0

    rank = p * (n - 1)
    lo = int(rank) # floor
    hi = lo + 1
    w = rank - lo # fraction

    if hi >= n: # p == 100%
        return float(a[-1])

    # interpolate
    return a[lo] * (1 - w) + a[hi] * w

print(quantile_calculator([3,1,4,1,5,9,2,6], 50)) # 3.5
print(quantile_calculator([10,20,30,40,50], 25)) # 20.0
print(quantile_calculator([10,20,30,40,50], 0)) # 10.0
print(quantile_calculator([10,20,30,40,50], 100)) # 50.0
```
