```python
from typing import List

def cumulative_product(nums: List[int]) -> List[int]:
    # edge: empty
    if not nums:
        return []

    result = []
    prod = 1
    for x in nums:
        prod *= x # handles 0, negatives automatically
        result.append(prod)
    return result

# quick test
print(cumulative_product([2,3,4])) # [2,6,24]
print(cumulative_product([])) # []
print(cumulative_product([5])) # [5]
print(cumulative_product([2,0,4])) # [2,0,0]
```

```python
# No special case for k==0 — prod starts at 1, first iteration does 1*2=2
# for x in nums not enumerate — you don't need index
# Clear names: prod, result
# Edge cases said out loud: empty, single element, zero, negative
# Complexity: O(n) time, O(n) space — if they ask O(1) extra, you can say "output doesn't count"
```

```python
def first_non_repeated(s: str) -> str:
  arr = list(s)
  result = ""

  if not arr:
    return result
    for i in range(len(arr)):
        if arr[i] not in arr[:i] and arr[i] not in arr[i + 1:]:
        # print(arr[i + 1:])
            result = arr[i]
            break
    return result

print(first_non_repeated("swiss"))
```
