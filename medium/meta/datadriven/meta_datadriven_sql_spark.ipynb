# Meta SQL / Spark — datadriven.io problem set

Working Spark SQL for all **120** problems in the Meta `SQL/Spark` track
([problem list](https://datadriven.io/problems?domain=sql%2Cpipeline_architecture%2Cai_coding&company=meta)),
each against a dataset built in this notebook. Difficulty mix: 72 x L3, 35 x L4, 11 x L5, 2 x L6.

**How it is put together**

1. A local `SparkSession` (section 0) — no cluster, no external data.
2. Section 1 registers every table as a temp view from in-notebook literals, so the whole
   notebook is reproducible offline and every query is plain `spark.sql(...)`.
3. Sections 2 onward: one cell per problem. Each header restates the task, then the query runs.

**Caveat worth stating up front:** the public problem list gives a title and a one-line teaser,
not the exact schema or expected output. Each task below is a faithful restatement of the
teaser against the schemas in section 1, and it is spelled out in the header so you can see
the interpretation. On the real grader, only the table and column names would change — the SQL
shape (window function, anti-join, grouping set, sweep line) is the part worth keeping.

**Spark version notes:** written for Spark 3.5. Spark SQL has no recursive CTE, so the two graph
problems (*Six Degrees*, *Cheapest CDN Route*) unroll the traversal to a fixed hop depth.

## 0. Spark session

Local Spark session used by every query below. All datasets are registered as temp views,
so every solution is plain `spark.sql(...)` — the same SQL you would run on a warehouse.

```python
import os
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("meta-datadriven-sql")
    .master("local[*]")
    .config("spark.sql.shuffle.partitions", "4")
    .config("spark.sql.session.timeZone", "UTC")
    .config("spark.driver.host", "127.0.0.1")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

def sql(q, n=50, truncate=False):
    """Run a query and show the result."""
    df = spark.sql(q)
    df.show(n, truncate=truncate)
    return df

print("Spark", spark.version)
```

## 1. Datasets

Twelve small, deterministic fact/dim sets covering the domains the problem set draws on:
product analytics, commerce, ads, notifications, messaging, infra/DevOps, Kubernetes,
streaming, source control, ML runs, feature flags and sensors.

Each table is deliberately seeded with the edge cases the problems hunt for: users with no
activity, duplicate keys, NULL keys, records spanning quarters, overlapping intervals.

```python
from datetime import date, datetime

def d(s):  return date.fromisoformat(s)
def ts(s): return datetime.fromisoformat(s)

def view(name, rows, schema):
    df = spark.createDataFrame(rows, schema)
    df.createOrReplaceTempView(name)
    print(f"{name:22s} {df.count():4d} rows")
    return df
```

### 21. Six Degrees

*Every reply ties two names together. Find whose web reaches the furthest.* — Sr. Data Engineer (L5)

**Task.** Treat follows as an undirected graph and expand up to six hops (Spark SQL has no recursive CTE, so the expansion is unrolled). Report each user's reachable set size and hop-6 radius.

```python
sql(r"""
WITH edges AS (
    SELECT user_id AS a, follows_id AS b FROM follows
    UNION
    SELECT follows_id AS a, user_id AS b FROM follows
),
h1 AS (SELECT a AS src, b AS dst, 1 AS hops FROM edges),
h2 AS (SELECT h1.src, e.b AS dst, 2 AS hops FROM h1 JOIN edges e ON e.a = h1.dst),
h3 AS (SELECT h2.src, e.b AS dst, 3 AS hops FROM h2 JOIN edges e ON e.a = h2.dst),
h4 AS (SELECT h3.src, e.b AS dst, 4 AS hops FROM h3 JOIN edges e ON e.a = h3.dst),
h5 AS (SELECT h4.src, e.b AS dst, 5 AS hops FROM h4 JOIN edges e ON e.a = h4.dst),
h6 AS (SELECT h5.src, e.b AS dst, 6 AS hops FROM h5 JOIN edges e ON e.a = h5.dst),
reach AS (
    SELECT src, dst, MIN(hops) AS hops
    FROM (SELECT * FROM h1 UNION ALL SELECT * FROM h2 UNION ALL SELECT * FROM h3
          UNION ALL SELECT * FROM h4 UNION ALL SELECT * FROM h5 UNION ALL SELECT * FROM h6)
    WHERE src <> dst
    GROUP BY src, dst
)
SELECT r.src AS user_id,
       u.name,
       COUNT(*)        AS people_reached_within_6,
       MAX(r.hops)     AS furthest_hop,
       ROUND(AVG(r.hops), 2) AS avg_degrees_of_separation
FROM reach r
JOIN users u ON u.user_id = r.src
GROUP BY r.src, u.name
ORDER BY people_reached_within_6 DESC, furthest_hop DESC, r.src
""")
```

### 23. The Quiet Outlier

*Ignore what the traffic does all day. Find the spike that barely showed up.* — Data Engineer (L4)

**Task.** A spike is an hour at least 3x the region's median volume. Among those, surface the one sitting in the quietest region — the spike that barely registers in absolute terms.

```python
sql(r"""
WITH med AS (
    SELECT region,
           percentile(requests, 0.5) AS median_requests,
           AVG(requests)             AS avg_requests
    FROM traffic_hourly
    GROUP BY region
), flagged AS (
    SELECT t.region,
           t.hour_ts,
           t.requests,
           m.median_requests,
           ROUND(t.requests / m.median_requests, 2) AS x_median
    FROM traffic_hourly t
    JOIN med m ON m.region = t.region
    WHERE t.requests >= 3 * m.median_requests
)
SELECT *,
       RANK() OVER (ORDER BY requests ASC) AS quietness_rank
FROM flagged
ORDER BY requests ASC
""")
```

*Every token lived among others. Find the moment the crowd was largest.* — Sr. Data Engineer (L5)

### 69. Bookends of the Week

*Every week has two edges. Weigh what lands on each.* — Data Engineer (L4)

**Task.** Per ISO week, how much activity landed on the week's two edges — Monday and Sunday — and which edge carried more.

```python
sql(r"""
WITH edges AS (
    SELECT weekofyear(session_date) AS iso_week,
           CASE date_format(session_date, 'E')
                WHEN 'Mon' THEN 'monday'
                WHEN 'Sun' THEN 'sunday'
           END AS edge,
           session_id,
           duration_sec
    FROM session_facts
    WHERE date_format(session_date, 'E') IN ('Mon', 'Sun')
)
SELECT iso_week,
       SUM(CASE WHEN edge = 'monday' THEN 1 ELSE 0 END)            AS monday_sessions,
       SUM(CASE WHEN edge = 'sunday' THEN 1 ELSE 0 END)            AS sunday_sessions,
       ROUND(SUM(CASE WHEN edge = 'monday'
                      THEN duration_sec ELSE 0 END) / 60.0, 1)     AS monday_minutes,
       ROUND(SUM(CASE WHEN edge = 'sunday'
                      THEN duration_sec ELSE 0 END) / 60.0, 1)     AS sunday_minutes,
       CASE WHEN SUM(CASE WHEN edge = 'monday' THEN duration_sec ELSE 0 END)
               > SUM(CASE WHEN edge = 'sunday' THEN duration_sec ELSE 0 END)
            THEN 'monday' ELSE 'sunday' END                        AS heavier_edge
FROM edges
GROUP BY iso_week
ORDER BY iso_week
""")
```

### 73. Diminishing Returns

*When search comes back thin, does the click still follow?* — Data Engineer (L4)

**Task.** Bucket searches by how many results came back, then measure whether clicks still follow.

```python
sql(r"""
WITH scored AS (
    SELECT s.search_id,
           s.results_count,
           CASE WHEN s.results_count = 0          THEN '0 results'
                WHEN s.results_count BETWEEN 1 AND 5  THEN '1-5 (thin)'
                WHEN s.results_count BETWEEN 6 AND 20 THEN '6-20'
                ELSE '20+' END AS result_bucket,
           COUNT(c.click_id)   AS clicks
    FROM searches s
    LEFT JOIN search_clicks c ON c.search_id = s.search_id
    GROUP BY s.search_id, s.results_count
)
SELECT result_bucket,
       COUNT(*)                                                        AS searches,
       SUM(clicks)                                                     AS clicks,
       SUM(CASE WHEN clicks > 0 THEN 1 ELSE 0 END)                     AS searches_with_click,
       ROUND(100.0 * SUM(CASE WHEN clicks > 0 THEN 1 ELSE 0 END)
             / COUNT(*), 1)                                            AS click_through_pct,
       ROUND(AVG(clicks), 2)                                           AS clicks_per_search
FROM scored
GROUP BY result_bucket
ORDER BY CASE result_bucket WHEN '0 results' THEN 0 WHEN '1-5 (thin)' THEN 1
                            WHEN '6-20' THEN 2 ELSE 3 END
""")
```

### 90. The Space Between

*The time between events is the tell.* — Data Engineer (L4)

**Task.** Per user, the gap between consecutive events: each gap plus the average and longest pause.

```python
sql(r"""
WITH gaps AS (
    SELECT user_id,
           event_time,
           LAG(event_time) OVER (PARTITION BY user_id ORDER BY event_time) AS prev_event_time,
           (unix_timestamp(event_time)
            - unix_timestamp(LAG(event_time) OVER (PARTITION BY user_id
                                                   ORDER BY event_time))) / 60.0 AS gap_min
    FROM events
    WHERE user_id IS NOT NULL
)
SELECT g.user_id,
       u.name,
       COUNT(g.gap_min)                  AS gaps,
       ROUND(AVG(g.gap_min), 1)          AS avg_gap_min,
       ROUND(MIN(g.gap_min), 1)          AS shortest_gap_min,
       ROUND(MAX(g.gap_min), 1)          AS longest_gap_min
FROM gaps g
JOIN users u ON u.user_id = g.user_id
WHERE g.gap_min IS NOT NULL
GROUP BY g.user_id, u.name
ORDER BY longest_gap_min DESC, g.user_id
""")
```

### 92. Daily Error Resolution Ratio

*Reported versus removed. The daily ratio.* — Data Engineer (L4)

**Task.** Errors reported versus errors resolved on each day, and the resolution ratio.

```python
sql(r"""
WITH reported AS (
    SELECT CAST(error_time AS DATE) AS day, COUNT(*) AS reported
    FROM errors
    GROUP BY CAST(error_time AS DATE)
), resolved AS (
    SELECT CAST(resolved_time AS DATE) AS day, COUNT(*) AS resolved
    FROM errors
    WHERE resolved_time IS NOT NULL
    GROUP BY CAST(resolved_time AS DATE)
)
SELECT COALESCE(rp.day, rs.day)                                          AS day,
       COALESCE(rp.reported, 0)                                          AS reported,
       COALESCE(rs.resolved, 0)                                          AS resolved,
       ROUND(COALESCE(rs.resolved, 0)
             / NULLIF(COALESCE(rp.reported, 0), 0), 2)                   AS resolution_ratio,
       COALESCE(rp.reported, 0) - COALESCE(rs.resolved, 0)               AS net_backlog_change
FROM reported rp
FULL OUTER JOIN resolved rs ON rs.day = rp.day
ORDER BY day
""")
```

### 110. Cheapest CDN Route

*The cheapest path across regions.* — Sr. Data Engineer (L5)

**Task.** Cheapest transfer route between regions, expanding the link graph up to four hops and keeping the minimum-cost path per origin/destination pair.

```python
sql(r"""
WITH l AS (
    SELECT from_region AS a, to_region AS b, cost_per_gb AS c FROM cdn_links
),
h1 AS (
    SELECT a AS src, b AS dst, c AS cost, CONCAT(a, ' -> ', b) AS path, 1 AS hops FROM l
),
h2 AS (
    SELECT h1.src AS src, l.b AS dst, h1.cost + l.c AS cost,
           CONCAT(h1.path, ' -> ', l.b) AS path, 2 AS hops
    FROM h1 JOIN l ON l.a = h1.dst
    WHERE INSTR(h1.path, l.b) = 0
),
h3 AS (
    SELECT h2.src AS src, l.b AS dst, h2.cost + l.c AS cost,
           CONCAT(h2.path, ' -> ', l.b) AS path, 3 AS hops
    FROM h2 JOIN l ON l.a = h2.dst
    WHERE INSTR(h2.path, l.b) = 0
),
h4 AS (
    SELECT h3.src AS src, l.b AS dst, h3.cost + l.c AS cost,
           CONCAT(h3.path, ' -> ', l.b) AS path, 4 AS hops
    FROM h3 JOIN l ON l.a = h3.dst
    WHERE INSTR(h3.path, l.b) = 0
),
all_paths AS (
    SELECT * FROM h1 UNION ALL SELECT * FROM h2
    UNION ALL SELECT * FROM h3 UNION ALL SELECT * FROM h4
),
ranked AS (
    SELECT src, dst, ROUND(cost, 4) AS cost, path, hops,
           ROW_NUMBER() OVER (PARTITION BY src, dst ORDER BY cost, hops) AS rnk
    FROM all_paths
)
SELECT src AS from_region, dst AS to_region, cost AS cheapest_cost_per_gb, hops, path
FROM ranked
WHERE rnk = 1
ORDER BY from_region, cheapest_cost_per_gb, to_region
""")
```
