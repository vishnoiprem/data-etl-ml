```python
def second_letter_counts(strings):
    d = {}
    for s in strings:
        w = s.split()
        if w and len(w[0]) > 1:
            c = w[0][1]
            d[c] = d.get(c, 0) + 1
    return d

# ---- test it ----
test = ["hello world", "hi there", "apple pie", "hello again", "a", ""]
result = second_letter_counts(test)

print(result)
```

```python
from collections import Counter

def second_letter_counts(strings):
    return dict(Counter(
        s.split()[0][1]
        for s in strings
        if s.split() and len(s.split()[0]) > 1
    ))

print(second_letter_counts(["hello world", "hi", "apple", "hello"]))
# {'e': 2, 'i': 1, 'p': 1}
```

```python
def second_highest_per_dept(employees):
    by_dept = {} # normal dict

    # 1. Group
    for e in employees:
        dept = e["dept"]
        if dept not in by_dept:
            by_dept[dept] = []
        by_dept[dept].append(e["salary"])

    # 2. Find 2nd highest
    result = {}
    for dept, salaries in by_dept.items():
        unique = sorted(set(salaries), reverse=True)
        if len(unique) >= 2:
            result[dept] = unique[1]
        else:
            result[dept] = None

    return result

# TEST
employees = [
    {"dept": "Eng", "salary": 100},
    {"dept": "Eng", "salary": 200},
    {"dept": "Eng", "salary": 300},
    {"dept": "HR", "salary": 500},
    {"dept": "HR", "salary": 400},
]

print(second_highest_per_dept(employees))
# {'Eng': 200, 'HR': 400}
```

```python
def remove_by_key(items, key, value):
    d=[]
    for x in items:
        if x.get(key) != value:
            d.append(x)
    return x

# TEST
data = [
    {"id": 1, "name": "a"},
    {"id": 2, "name": "b"},
    {"id": 1, "name": "c"},
]

new_data = remove_by_key(data, "id", 1)
print(new_data)
# [{'id': 2, 'name': 'b'}]
```
