```python
def activity_time_ledger(activities: list[dict], targets: list[str]) -> int:

  output = 0

  for data in activities:
    name = data['name']
    minutes = data['minutes']
    for t in targets:
      if t == name:
        output = output+minutes
  return output




# https://datadriven.io/problems/the_ones_that_count

# targets
# :
# ["run","code"]
# activities
# :
# [
#   {"name":"run","minutes":30},
#   {"name":"read","minutes":45},
#   {"name":"code","minutes":60},
#   {"name":"run","minutes":20}
# ]





```

```python
def load_balancer_round_robin(servers: list, requests: list) -> dict:
  d = {}

  if not servers:
    return d

  n = len(servers)

  for i, request in enumerate(requests):
    d[request] = servers[i % n]

  return d


# https://datadriven.io/problems/the_traffic_director


# servers
# :
# ["s1","s2","s3"]
# requests
# :
# ["r1","r2","r3","r4"]
# Output

```

```python
def parse_csv_rows(lines):
    result = []

    if not lines:
        return result

    headers = lines[0].split(",")

    for line in lines[1:]:
        values = line.split(",")
        d = {}

        for i in range(len(headers)):
            d[headers[i]] = values[i]

        result.append(d)

    result.sort(key=lambda x: x[headers[0]])

    return result
```

```python
def word_counts(text: str) -> dict:
  result = {}

  if not text:
    return {}

  temp = text.split()  # Fixed

  for row in temp:
    if row in result:
      result[row] += 1
    else:
      result[row] = 1

  return dict(sorted(result.items(), key=lambda x: x[1], reverse=True))
```
