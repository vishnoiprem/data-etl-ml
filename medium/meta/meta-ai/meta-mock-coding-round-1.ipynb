```python

from collections import Counter

def top_entities(interactions, n=10):
    # interactions = list of entity_ids
    cnt = Counter(interactions)
    return cnt.most_common(n)
    # O(N) time, O(K) space
```

```python
from datetime import datetime, timedelta

def avg_requests(requests):
    # requests: list of dicts {user_id, requested_at: datetime}
    cutoff = datetime.now() - timedelta(days=7)
    weekly=[]
    weekly = for r in requests:
                if r['requested_at'] >= cutoff:
                    weekly.append(r)
    
    from collections import defaultdict
    counts = defaultdict(int)
    for r in weekly:
        counts[r['user_id']] += 1
    
    if not counts: return 0
    return sum(counts.values()) / len(counts)

```
