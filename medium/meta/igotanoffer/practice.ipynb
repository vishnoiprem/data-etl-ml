```python
def freind_shipt(pari):

    result = {}

    for row in pari:
        for col in row:
            print(col, end=' ')
            if col not in result:
                result[col] = 0
            result[col] += 1

    for person in pari:
        if isinstance(person, list):
            for p in person:
                if p not in result:
                    result[p] = 0
        else:
            if person not in result:
                result[person] = 0
    return  result
data=[[2,3], [3,4], [5]]

print(freind_shipt(data))
```

```python
def count_friends(friendships):
    result = {}

    # For each friendship pair
    for pair in friendships:
        for person in pair:
            if person not in result:
                result[person] = 0
            result[person] += 1

    # Make sure person 5 is included (0 friends)
    for person in friendships:
        if isinstance(person, list):
            for p in person:
                if p not in result:
                    result[p] = 0
        else:
            if person not in result:
                result[person] = 0

    return result

data=[[2,3], [3,4], [5]]

print(freind_shipt(data))
```

```python
from collections import defaultdict, Counter
def friends(pairs):
    c = Counter()
    for a, b in pairs:
        c[a] += 1; c[b] += 1
    return dict(c)

nums = [[2, 3], [3, 4], [5]]                                                                                                                                                                                                     
print(friend(nums))
```
