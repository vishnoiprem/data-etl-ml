```python
def smallest_from_digits(digits):
    digits.sort()
    if digits[0] == 0 and all(d==0 for d in digits):
        return 0

    # find first non-zero
    for i in range(len(digits)):
        if digits[i]!= 0:
            # swap it to front
            digits[0], digits[i] = digits[i], digits[0]
            break

    # make number
    return int(''.join(map(str, digits)))

# TEST
print(smallest_from_digits([0,1,0,2,3])) # 10023
print(smallest_from_digits([0,0,0,1])) # 1000
```

```python
# 23Max consecutive workshops	Sort + adjacent sum

def max_consecutive(arr):
    arr.sort() # O(n log n)
    max_sum = 0
    for i in range(len(arr)-1):
        s = arr[i] + arr[i+1]
        if s > max_sum:
            max_sum = s
    return max_sum

print(max_consecutive([4,2,7,1])) # 11
# 1,2,4,7,
print(max_consecutive([5,5,5,5])) # 10
```

```python

# 22.Largest odd	Sort odds/evens separately, place largest odd last
# Given numbers, separate odd and even. Sort both. Put biggest odd at the very end.

def largest_odd_last(arr):
    odds = sorted([x for x in arr if x % 2 == 1])
    evens = sorted([x for x in arr if x % 2 == 0])
    
    if not odds:
        return evens # no odd present
    
    return evens + odds # largest odd is odds[-1], already last

print(largest_odd_last([5,2,9,4,1,8])) # [2,4,8,1,5,9]
print(largest_odd_last([4,1,2,3])) # [2,4,1,3]
```

```python
def most_frequent_dedup(books):
    """
    books = {"bookA": ["good","good","bad"],...}
    returns (review, count)
    """
    freq = {}

    for reviews in books.values():
        # dedup per book
        seen = set()
        for r in reviews:
            if r in seen:
                continue
            seen.add(r)
            freq[r] = freq.get(r, 0) + 1

    # find max
    best_review = None
    best_count = 0
    for review, count in freq.items():
        if count > best_count:
            best_count = count
            best_review = review

    return best_review, best_count

# TEST
books = {
  "bookA": ["good", "good", "bad"],
  "bookB": ["bad", "good"],
  "bookC": ["good", "excellent"]
}

print(most_frequent_dedup(books))

# ('good', 3)
```

```python
def most_frequent_review(books):
    freq = {}
    for reviews in books.values():
        for r in reviews: # no set() here
            freq[r] = freq.get(r, 0) + 1

    best_review = None
    best_count = 0
    for review, count in freq.items():
        if count > best_count:
            best_count = count
            best_review = review

    return best_review, best_count

books = {
  "bookA": ["good", "good", "bad", "good"],
  "bookB": ["bad", "bad", "good"],
}
print(most_frequent_review(books)) # ('good', 4)
```

```python
def find_sequels(titles):
    groups = {}
    for title in titles:
        first = title.split()[0]
        if first not in groups:
            groups[first] = []
        groups[first].append(title)

    # only sequels where count > 1
    sequels = {}
    for first, lst in groups.items():
        if len(lst) > 1:
            sequels[first] = lst
    return sequels

# TEST
titles = [
  "Harry Potter 1",
  "Harry Potter 2",
  "Lord of Rings",
  "Harry Chamber",
  "Lord Hobbit"
]

print(find_sequels(titles))
# {'Harry': ['Harry Potter 1','Harry Potter 2','Harry Chamber'],
# 'Lord': ['Lord of Rings','Lord Hobbit']}
```

```python
def max_books(prices, budget):
    prices.sort() # cheapest first
    count = 0
    total = 0
    print(prices)
    for p in prices:
        if total + p <= budget:
            total += p
            count += 1
        else:
            break
    return count

print(max_books([10,20,5,30], 35)) # 3
print(max_books([5,5,5], 10)) # 2
```

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False

class SpellChecker:
    def __init__(self):
        self.root = TrieNode()

    def add(self, word):
        node = self.root
        for ch in word.lower():
            if ch not in node.children:
                node.children[ch] = TrieNode()
            node = node.children[ch]
        node.is_end = True

    def check(self, word):
        node = self.root
        for ch in word.lower():
            if ch not in node.children:
                return False
            node = node.children[ch]
        return node.is_end

checker = SpellChecker()
for w in ["apple","book","harry"]: checker.add(w)

print(checker.check("book")) # True
print(checker.check("bok")) # False
```

```python
def parse(tokens):
    # tokens = ["(", "Harry", "OR", "Lord", ")"]
    def parse_expr(pos):
        # yields (node, next_pos) for backtracking
        if tokens[pos] == "(":
            # try to parse inside
            for node, next_pos in parse_expr(pos+1):
                if next_pos < len(tokens) and tokens[next_pos] == ")":
                    # expect operator after
                    if next_pos+1 < len(tokens) and tokens[next_pos+1] in ("AND","OR"):
                        op = tokens[next_pos+1]
                        for right_node, final_pos in parse_expr(next_pos+2):
                            yield ((op, node, right_node), final_pos)
                    else:
                        yield (node, next_pos+1)
        else:
            # base case - single word
            yield (tokens[pos], pos+1)

    # try all parses
    for result, pos in parse_expr(0):
        if pos == len(tokens):
            return result
    return None

print(parse(["(", "Harry", "OR", "Lord", ")"]))
# ('OR', 'Harry', 'Lord')

print(parse(["Harry"]))
# Harry
```

```python
def flatten_dict_no_yield(d, parent="", result=None):
    if result is None:
        result = {}
    for k, v in d.items():
        new_key = f"{parent}.{k}" if parent else k
        if isinstance(v, dict):
            flatten_dict_no_yield(v, new_key, result)
        else:
            result[new_key] = v
    return result

d = {
  "book": {"price": {"usd": 10}},
  "name": "Harry"
}

print(flatten_dict_no_yield(d))
# {'book.price.usd': 10, 'name': 'Harry'}"
```
