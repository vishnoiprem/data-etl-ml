```python
def freind_shipt(pari):

    result = {}

    for row in pari:
        for col in row:
            print(col, end=' ')
            if col not in result:
                result[col] = 0
            result[col] += 1

    for person in pari:
        if isinstance(person, list):
            for p in person:
                if p not in result:
                    result[p] = 0
        else:
            if person not in result:
                result[person] = 0
    return  result
data=[[2,3], [3,4], [5]]

print(freind_shipt(data))
```
